```python
df.columns = (
    df.columns
    .str.strip()
    .str.replace("\n", " ", regex=False)
    .str.replace(r"\s+", " ", regex=True)
)
```

```python
print(df.columns.tolist())
```

```python
column_mapping = {
    "Timestamp": "Timestamp",
    "Education level": "Education_Level",
    "Age group": "Age_Group",
    "Gender": "Gender",
    "Occupation": "Occupation",
    "Monthly household income (INR):": "Income_Group",
    "City category": "City_Category",
    "Have you watched a movie in a cinema theatre in the past 12 months?": "Cinema_Visit",
    "Do you currently use any OTT/video streaming platforms (Netflix, Amazon Prime, Hotstar, etc.)?": "OTT_User",
    "Do you currently have any OTT subscription?": "OTT_Subscription",
    "How many hours per week do you watch OTT content?": "OTT_Hours",
    "I prefer watching new movies at home rather than going to theatres": "Home_Movie_Preference",
    "Watching in theatres feels like a special experience worth paying for.": "Cinema_Experience",
    "OTT provides better convenience and variety than theatres.": "OTT_Convenience",
    "I am willing to pay extra for early OTT release or premium experience.": "Premium_OTT_WTP",
    "Expensive tickets stop me from visiting theatres frequently.": "Ticket_Price_Barrier",
    "If a new blockbuster released today, which would you choose?": "Blockbuster_Choice",
    "If OTT prices increased by 25%, would you continue your subscription?": "OTT_Price_Response",
    "How long should the gap be between theatre and OTT release?": "Release_Gap",
    "What makes you decide between watching a movie in theatre vs OTT?": "Decision_Factor"
}

df = df.rename(columns=column_mapping)

df.head()
```

Check missing values

```python
missing = pd.DataFrame({
    "Missing_Count": df.isna().sum(),
    "Missing_Percentage": (df.isna().mean() * 100).round(2)
})

missing = missing[missing["Missing_Count"] > 0]

missing.sort_values("Missing_Count", ascending=False)
```

Handle categorical missing values

```python
categorical_cols = [
    "Education_Level",
    "Age_Group",
    "Gender",
    "Occupation",
    "Income_Group",
    "City_Category",
    "Cinema_Visit",
    "OTT_User",
    "OTT_Subscription",
    "OTT_Hours",
    "Blockbuster_Choice",
    "OTT_Price_Response",
    "Release_Gap"
]

for col in categorical_cols:
    df[col] = df[col].fillna("Not Specified")
```

Handle the open-ended question

```python
df["Decision_Factor"] = df["Decision_Factor"].fillna("")   #Decision_Factor
```

Standardize Yes/No variables

```python
yes_no_cols = [
    "Cinema_Visit",
    "OTT_User",
    "OTT_Subscription"
]

for col in yes_no_cols:
    df[col] = (
        df[col]
        .astype(str)
        .str.strip()
        .str.lower()
        .replace({
            "yes": "Yes",
            "no": "No"
        })
    )
```

```python
for col in yes_no_cols:
    print("\n", col)
    print(df[col].value_counts())
```

Clean OTT Hours

```python
ott_hours_mapping = {
    "<1 hour": 0.5,
    "1–3 hours": 2,
    "4–7 hours": 5.5,
    "8–14 hours": 11,
    "15+ hours": 15
}

df["OTT_Hours_Numeric"] = df["OTT_Hours"].map(ott_hours_mapping)  # for categorical analysis
```

Check Likert variables

```python
likert_cols = [
    "Home_Movie_Preference",
    "Cinema_Experience",
    "OTT_Convenience",
    "Premium_OTT_WTP",
    "Ticket_Price_Barrier"
]
```

```python
for col in likert_cols:
    print("\n", col)
    print(df[col].value_counts().sort_index())
```

```python
for col in likert_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")
```

```python
df[likert_cols].describe().T
```

Validate Likert values

```python
for col in likert_cols:
    invalid = df.loc[
        ~df[col].isin([1, 2, 3, 4, 5]) & df[col].notna(),
        col
    ]
    
    print(col, "Invalid values:", invalid.tolist())
```

Clean Blockbuster Choice

```python
df["Blockbuster_Choice"].value_counts(dropna=False)
```

```python
def simplify_choice(x):
    if pd.isna(x):
        return np.nan
    
    x = str(x).lower()
    
    if "cinema" in x:
        return "Cinema"
    elif "wait" in x and "ott" in x:
        return "OTT"
    elif "499" in x or "same day" in x:
        return "PVOD"
    else:
        return "Other"

df["Choice_Simple"] = df["Blockbuster_Choice"].apply(simplify_choice)
```

```python
df["Choice_Simple"].value_counts(dropna=False)
```

Create a clean demographic table

```python
demographic_cols = [
    "Education_Level",
    "Age_Group",
    "Gender",
    "Occupation",
    "Income_Group",
    "City_Category"
]

for col in demographic_cols:
    print("\n==========", col, "==========")
    print(df[col].value_counts(dropna=False))
```

```python
df["Gender"] = (
    df["Gender"]
    .str.strip()
    .str.title()
)

df["Occupation"] = (
    df["Occupation"]
    .str.strip()
    .str.title()
)
```

Timestamp cleaning

```python
df["Timestamp"] = pd.to_datetime(
    df["Timestamp"],
    errors="coerce"
)
```

```python
df["Response_Date"] = df["Timestamp"].dt.date
df["Response_Hour"] = df["Timestamp"].dt.hour
```

Creating the first derived variables

```python
df["OTT_Engagement_Score"] = (
    df["OTT_Hours_Numeric"]
)
```

```python
df["Cinema_Experience_Score"] = df["Cinema_Experience"]
```

```python
df["Price_Sensitivity_Score"] = df["Ticket_Price_Barrier"]
```
