### Cinema vs OTT vs PVOD

```python
df["Choice_Simple"].value_counts()
```

```python
# Percentage 

choice_pct = (
    df["Choice_Simple"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

print(choice_pct)
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    data=df,
    y="Choice_Simple",
    order=df["Choice_Simple"].value_counts().index
)

plt.title("Preferred Platform for a New Blockbuster")
plt.xlabel("Number of Respondents")
plt.ylabel("Choice")

plt.tight_layout()
plt.show()
```

Gender × Cinema/OTT

```python
gender_choice = pd.crosstab(
    df["Gender"],
    df["Choice_Simple"]
)

gender_choice
```

```python
# Convert it to percentages

gender_choice_pct = pd.crosstab(
    df["Gender"],
    df["Choice_Simple"],
    normalize="index"
) * 100

gender_choice_pct.round(2)
```

```python
gender_choice_pct.plot(
    kind="bar",
    stacked=True,
    figsize=(9,6)
)

plt.title("Cinema vs OTT Preference by Gender")
plt.xlabel("Gender")
plt.ylabel("Percentage of Respondents")

plt.legend(title="Choice")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

### Age × Cinema/OTT

```python
age_choice = pd.crosstab(
    df["Age_Group"],
    df["Choice_Simple"],
    normalize="index"
) * 100

age_choice.round(2)
```

```python
age_choice.plot(
    kind="bar",
    stacked=True,
    figsize=(10,6)
)

plt.title("Cinema vs OTT Preference by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Percentage of Respondents")

plt.legend(title="Choice")

plt.tight_layout()
plt.show()
```

OTT Usage × Preference

```python
ott_choice = pd.crosstab(
    df["OTT_Hours"],
    df["Choice_Simple"],
    normalize="index"
) * 100

ott_choice.round(2)
```

```python
ott_choice.plot(
    kind="bar",
    stacked=True,
    figsize=(10,6)
)

plt.title("Blockbuster Preference by OTT Viewing Intensity")
plt.xlabel("Weekly OTT Viewing")
plt.ylabel("Percentage of Respondents")

plt.xticks(rotation=30)
plt.legend(title="Choice")

plt.tight_layout()
plt.show()
```

## Likert Analysis

```python
likert_cols = [
    "Home_Movie_Preference",
    "Cinema_Experience",
    "OTT_Convenience",
    "Premium_OTT_WTP",
    "Ticket_Price_Barrier"
]
```

```python
likert_summary = df[likert_cols].describe().T

likert_summary[
    ["count", "mean", "std", "min", "max"]
]
```

Mean Likert comparison

```python
mean_scores = df[likert_cols].mean().sort_values()

plt.figure(figsize=(9,6))

mean_scores.plot(kind="barh")

plt.title("Average Consumer Attitude Scores")
plt.xlabel("Mean Likert Score")
plt.ylabel("Variable")

plt.xlim(1,5)

plt.tight_layout()
plt.show()
```

# Correlation Analysis

```python
correlation = df[likert_cols].corr(method="spearman")

correlation.round(2)
```

```python
plt.figure(figsize=(10,7))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0
)

plt.title("Spearman Correlation Between Consumer Attitudes")

plt.tight_layout()
plt.show()
```
